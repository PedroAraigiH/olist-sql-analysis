```python
run(f"""
WITH monthly_revenue AS (
    SELECT
        strftime('%Y-%m', order_purchase_timestamp) AS month,
        COUNT(DISTINCT i.order_id) AS n_orders,
        SUM(price) AS revenue
    FROM orders o
    JOIN order_items i ON o.order_id = i.order_id
    WHERE {SCOPE}
    GROUP BY month
),
revenue_growth AS (
    SELECT
        month,
        n_orders,
        revenue,
        LAG(revenue) OVER (ORDER BY month) AS previous_month_revenue
    FROM monthly_revenue
)
SELECT
    month,
    n_orders,
    revenue,
    previous_month_revenue,
    ROUND(
        (revenue - previous_month_revenue) * 100.0
        / NULLIF(previous_month_revenue, 0),
        2
    ) AS growth_pct
FROM revenue_growth
ORDER BY month;
""")
```

```python
run(f"""
WITH monthly_revenue AS (
    SELECT 
        strftime('%Y-%m', order_purchase_timestamp) AS month, 
        COUNT(DISTINCT i.order_id) AS n_orders, 
        SUM(price) AS revenue
    FROM orders o 
    JOIN order_items i ON o.order_id=i.order_id 
    WHERE {SCOPE}
    GROUP BY month
)
SELECT 
    month,
    revenue,
    SUM(revenue) OVER (ORDER BY month) AS cumulative_revenue,
    ROUND(SUM(revenue) OVER (ORDER BY month)/SUM(revenue) OVER () *100, 2) AS cumulative_share
FROM monthly_revenue;
""")
```

```python
run(f"""
WITH seller_category AS (
    SELECT COALESCE(t.product_category_name_english, p.product_category_name, 'unknown') AS category,
           i.seller_id,
           SUM(i.price) AS revenue
    FROM order_items i 
    JOIN orders o ON i.order_id= o.order_id 
    JOIN products p ON i.product_id = p.product_id 
    LEFT JOIN product_category_name_translation t ON p.product_category_name = t.product_category_name 
    WHERE {SCOPE}
    GROUP BY category, i.seller_id
),
ranked AS (
    SELECT category, seller_id, revenue,
           ROW_NUMBER() OVER (PARTITION BY category ORDER BY revenue DESC) AS rank_in_category,
           SUM(revenue) OVER (PARTITION BY category) AS category_revenue
    FROM seller_category
)
SELECT category, seller_id, ROUND(revenue, 2) AS revenue, rank_in_category,
       ROUND(100.0 * revenue / category_revenue, 1) AS share_pct
FROM ranked
WHERE rank_in_category <= 3
  AND category IN ('health_beauty', 'watches_gifts', 'bed_bath_table', 'sports_leisure', 'computers_accessories')
ORDER BY category, rank_in_category;
""")
```

```python
run(f"""
WITH customer_orders AS (
    SELECT c.customer_unique_id, COUNT(DISTINCT o.order_id) AS n_orders
    FROM orders o
    JOIN customers c ON o.customer_id = c.customer_id
    WHERE {SCOPE}
    GROUP BY c.customer_unique_id
)
SELECT CASE WHEN n_orders >= 3 THEN '3+' ELSE CAST(n_orders AS TEXT) END AS orders_placed,
       COUNT(*) AS n_customers,
       ROUND(100.0 * COUNT(*) / SUM(COUNT(*)) OVER (), 2) AS pct_customers
FROM customer_orders
GROUP BY orders_placed
ORDER BY orders_placed;
""")
```
